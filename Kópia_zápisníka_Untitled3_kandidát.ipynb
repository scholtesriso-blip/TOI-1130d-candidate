```python

import lightkurve as lk
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# TOI-178 je TIC 251848941 - ideme cez TIC, stabilnejsie
search = lk.search_lightcurve("TIC 251848941", mission="TESS", author="SPOC")
print(f"Našiel {len(search)} sektorov")

# Stiahni len SPOC 2-min
lc = search.download_all().stitch().remove_nans().normalize()
lc = lc.remove_outliers(sigma=6)
print(f"TOI-178: {len(lc.time)} bodov")

pg = lc.to_periodogram(method='bls', period=np.linspace(2, 25, 3000), duration=0.12)
pg.plot()
plt.show()

best = pg.period_at_max_power
print(f"NAJLEPSI: {best:.4f} d sila {pg.power.max():.1f}")
lc.fold(period=best).scatter(s=1)
plt.title(f"TOI-178 {best:.4f} d")
plt.show()
```

```python

# LOV 7. PLANETY - TOI-178
lc_flat = lc.flatten(window_length=1001)
pg_new = lc_flat.to_periodogram(method='bls', period=np.linspace(2.5, 35, 5000), duration=0.2)
pg_new.plot()
plt.show()

# najdi top mimo znamych
best_periods = []
for i in np.argsort(pg_new.power.value)[::-1][:30]:
    p = pg_new.period.value[i]
    if any(abs(p-k)<0.5 for k in [1.91, 3.23, 6.55, 9.96, 15.23, 20.71]): continue
    if any(abs(p-k/2)<0.3 for k in [1.91, 3.23, 6.55, 9.96, 15.23, 20.71]): continue
    best_periods.append((p, pg_new.power.value[i]))
    if len(best_periods)>=5: break

print("KANDIDATI NA 7. PLANETU:")
for p,s in best_periods:
    print(f"{p:.4f} d sila {s:.1f}")
    lc_flat.fold(period=p).scatter(s=1)
    plt.title(f"Kandidat {p:.3f} d")
    plt.xlim(-0.3, 0.3)
    plt.ylim(0.995, 1.005)
    plt.show()
```

```python

import lightkurve as lk, numpy as np, matplotlib.pyplot as plt
import warnings; warnings.filterwarnings("ignore")

search = lk.search_lightcurve("TOI-500", mission="TESS")
print(f"TOI-500: {len(search)} sektorov total, beriem len 8")
search_small = search[:8]  # len 8 ako pri TOI-178

lcs = []
for i, s in enumerate(search_small):
    try:
        lc = s.download(quality_bitmask='hard')
        lcs.append(lc)
        print(f"{i+1}/8 ok")
    except:
        print(f"{i+1}/8 fail - preskakujem")
        continue

lc = lk.LightCurveCollection(lcs).stitch().remove_nans().normalize().remove_outliers(sigma=6)
print(f"MAME: {len(lc.time)} bodov - IDEME LOVIT!")

lc_flat = lc.flatten(window_length=1001)
pg = lc_flat.to_periodogram(method='bls', period=np.linspace(2, 35, 4000), duration=0.18)
pg.plot(); plt.show()
print(f"NAJLEPSI: {pg.period_at_max_power:.4f} d sila {pg.power.max():.1f}")
```

```python

# TOI-1130 - NEPREBADANY, 2 PLANETY, SANCA NA 3.
import lightkurve as lk
search = lk.search_lightcurve("TOI-1130", mission="TESS")
print(f"TOI-1130: {len(search)} sektorov")
s_small = search[:8]

lcs = []
for i,s in enumerate(s_small):
    try:
        lc_t = s.download(quality_bitmask='hard')
        lcs.append(lc_t)
        print(f"{i+1}/8 ok")
    except:
        print(f"{i+1} fail")
        continue

lc2 = lk.LightCurveCollection(lcs).stitch().remove_nans().normalize().remove_outliers(sigma=5)
lc2_flat = lc2.flatten(window_length=1001)
print(f"MAME {len(lc2.time)} bodov TOI-1130")

pg3 = lc2_flat.to_periodogram(method='bls', period=np.linspace(2, 20, 4000), duration=0.15)
pg3.plot(); import matplotlib.pyplot as plt; plt.show()
print(f"TOP: {pg3.period_at_max_power:.4f} d sila {pg3.power.max():.1f}")

# Rovno fold top 3
import numpy as np
for i in np.argsort(pg3.power.value)[::-1][:3]:
    p = pg3.period.value[i]
    if abs(p-4.07)<0.5 or abs(p-8.35)<0.5: continue
    lc2_flat.fold(period=p).scatter(s=1)
    plt.xlim(-0.25,0.25); plt.ylim(0.996,1.004)
    plt.title(f"TOI-1130 KANDIDAT {p:.4f} d sila {pg3.power.value[i]:.1f}")
    plt.show()
```

```python

# OPRAVA - flatten po sektoroch, nie po stitchi
import lightkurve as lk, matplotlib.pyplot as plt
import numpy as np

# uz mas lcs z predchadzajucej bunky - 8 sektorov TOI-1130
# flatten kazdy sektor zvlast, aby nezmazal tranzity cez medzery
lcs_flat = []
for lc in lcs:
    try:
        c = lc.remove_nans().normalize().remove_outliers(sigma=6)
        f = c.flatten(window_length=501, break_tolerance=5) # 501 = kratsie, break = nespajat cez medzery
        lcs_flat.append(f)
    except:
        continue

lc2_flat_fixed = lk.LightCurveCollection(lcs_flat).stitch()
print(f"FIXED: {len(lc2_flat_fixed.time)} bodov, po sektorovom flattnene")

# Teraz fold ZNAME - musia ukazat dieru!
for p_known in [4.074, 8.350]:
    fold = lc2_flat_fixed.fold(period=p_known)
    plt.figure(figsize=(8,3))
    fold.scatter(s=1, c='gray', alpha=0.3)
    binned = fold.bin(time_bin_size=0.03)
    binned.scatter(s=40, c='red')
    plt.xlim(-0.2, 0.2)
    plt.ylim(0.992, 1.008)
    plt.axvline(0, c='black', ls='--')
    plt.title(f"FIXED TOI-1130 ZNAMA {p_known} d - TERAZ MUSI IST DOLE!")
    plt.show()

# A teraz HLADAJ 3. po oprave
pg_fixed = lc2_flat_fixed.to_periodogram(method='bls', period=np.linspace(2.5, 25, 5000), duration=0.2)
pg_fixed.plot(); plt.show()

print(f"TOP po oprave: {pg_fixed.period_at_max_power:.4f} d sila {pg_fixed.power.max():.1f}")

best = []
for i in np.argsort(pg_fixed.power.value)[::-1][:15]:
    p = pg_fixed.period.value[i]
    if abs(p-4.074)<0.4 or abs(p-8.350)<0.6: continue
    best.append(p)
    if len(best)>=1: break

if best:
    p = best[0]
    fold = lc2_flat_fixed.fold(period=p)
    binned = fold.bin(time_bin_size=0.03)
    plt.figure(figsize=(8,4))
    fold.scatter(s=1, c='gray', alpha=0.2)
    binned.scatter(s=40, c='red')
    plt.xlim(-0.25,0.25); plt.ylim(0.993,1.007)
    plt.title(f"OPRAVENY KANDIDAT 3. PLANETY {p:.4f} d")
    plt.show()
    print(f"KANDIDAT: {p:.4f} d - ak cervena dole pod 0.998, MAME JU!")
```

```python

# FINAL FIX - pouzi transit_time z BLS!
import matplotlib.pyplot as plt

# zober t0 z periodogramu
p_best = pg_fixed.period_at_max_power.value
t0_best = pg_fixed.transit_time_at_max_power.value
print(f"Najsilnejsi: {p_best:.4f} d t0 {t0_best:.2f}")

# Fold ZNAME so spravnym t0
for p_known in [4.074, 8.350]:
    # najdi najblizsi t0 pre tuto periodu
    idx = (abs(pg_fixed.period.value - p_known)).argmin()
    t0 = pg_fixed.transit_time.value[idx]
    fold = lc2_flat_fixed.fold(period=p_known, epoch_time=t0)
    binned = fold.bin(time_bin_size=0.03)
    plt.figure(figsize=(8,4))
    fold.scatter(s=1, c='gray', alpha=0.2)
    binned.scatter(s=50, c='red')
    plt.xlim(-0.25,0.25); plt.ylim(0.988,1.006)
    plt.axvline(0, c='black', ls='--')
    plt.title(f"TOI-1130 ZNAMA {p_known} d s t0 - TERAZ!")
    plt.show()
    print(f"{p_known} d dip: {binned.flux.min():.5f}")

# Kandidat 16.7 so spravnym t0
p = 16.7003
idx = (abs(pg_fixed.period.value - p)).argmin()
t0 = pg_fixed.transit_time.value[idx]
fold = lc2_flat_fixed.fold(period=p, epoch_time=t0)
binned = fold.bin(time_bin_size=0.03)
plt.figure(figsize=(8,4))
fold.scatter(s=1, c='gray', alpha=0.2)
binned.scatter(s=50, c='red')
plt.xlim(-0.3,0.3); plt.ylim(0.985,1.007)
plt.axvline(0, c='black', ls='--')
plt.title(f"KANDIDAT 3. PLANETA {p:.4f} d - REAL?")
plt.show()
print(f"Kandidat dip: {binned.flux.min():.5f} - pod 0.995 = REAL!")
```

```python

# HLADAME REALNU 3. PLANETU - bez 2x, 3x, 0.5x
import numpy as np, matplotlib.pyplot as plt

def is_alias(p, known=[4.074, 8.35], tol=0.4):
    for k in known:
        for mult in [0.5, 1.0, 2.0, 3.0, 4.0]:
            if abs(p - k*mult) < tol:
                return True
            if abs(p - k/mult) < tol:
                return True
    return False

candidates = []
for i in np.argsort(pg_fixed.power.value)[::-1]:
    p = pg_fixed.period.value[i]
    pw = pg_fixed.power.value[i]
    if p < 3 or p > 30: continue
    if is_alias(p): continue
    if pw < 100: continue
    candidates.append((p, pw, pg_fixed.transit_time.value[i]))
    if len(candidates)>=6: break

print("TOP 6 REALNYCH (nie alias) KANDIDATOV TOI-1130:")
for p,pw,t0 in candidates:
    print(f"{p:.4f} d sila {pw:.0f} t0 {t0:.2f}")
    fold = lc2_flat_fixed.fold(period=p, epoch_time=t0)
    binned = fold.bin(time_bin_size=0.04)
    plt.figure(figsize=(8,4))
    fold.scatter(s=1, c='gray', alpha=0.15)
    binned.scatter(s=50, c='red')
    plt.xlim(-0.3,0.3); plt.ylim(0.989,1.006)
    plt.axvline(0, c='black', ls='--')
    plt.title(f"REAL KANDIDAT {p:.4f} d sila {pw:.0f} - NIE 2x!")
    plt.show()
    print(f"dip: {binned.flux.min():.5f} - ak pod 0.998 a U tvar, FOTIME!")
```

```python

# Bunka 13 - velkost kandidata 11.13 d
import numpy as np

dip = 0.99780  # z tvojho grafu
depth = 1 - dip  # 0.00220
rp_rs = np.sqrt(depth)
print(f"Hĺbka: {depth*100:.3f}% = {depth*100*10000:.0f} ppm")
print(f"Rp/Rs = sqrt(depth) = {rp_rs:.4f}")

# TOI-1130 hviezda: K6V, Rs ~ 0.71 Rsun, 154 ly (z ExoFOP)
Rs_sun = 0.71
Rs_earth = Rs_sun * 109.2  # Rsun v Rearth
Rp_earth = rp_rs * Rs_earth

print(f"Predpoklad Rs = {Rs_sun} Rsun")
print(f"Rp = {Rp_earth:.2f} Rearth = {Rp_earth/11.2:.2f} Rjup")

if Rp_earth < 1.6:
    typ = "Super-Zem"
elif Rp_earth < 4:
    typ = "Sub-Neptune - ako TOI-1130 b!"
elif Rp_earth < 8:
    typ = "Neptune"
else:
    typ = "Jupiter"

print(f"Typ: {typ}")

# Perióda 11.13 d - teplota
# a = (P/365)^2/3 pre 1 Msun, ale pre 0.7 Msun mensie
print(f"\nKandidát TOI-1130 d (?) : P=11.1327 d, Rp~{Rp_earth:.1f} Rearth")
print("To by bola 3. planéta medzi b(4.07) a c(8.35)? Nie, za nimi! Vonkajšia!")
print("4.07 - 8.35 - 11.13 = takmer 1:2:2.7 - blízko rezonancie!")
```

```python

# Bunka 14 FIX - VETTING 11.13 d bez chyby
import numpy as np, matplotlib.pyplot as plt

p = 11.1327
idx = (abs(pg_fixed.period.value - p)).argmin()
t0 = pg_fixed.transit_time.value[idx]
print(f"Vetting pre P={p:.4f} d t0={t0:.4f}")

fold = lc2_flat_fixed.fold(period=p, epoch_time=t0)

n_transit = np.floor((fold.time.value - t0)/p + 0.5)
odd_mask = (n_transit % 2 == 0)
even_mask = (n_transit % 2 == 1)

fold_odd = fold[odd_mask]
fold_even = fold[even_mask]

b_odd = fold_odd.bin(time_bin_size=0.04).flux.min()
b_even = fold_even.bin(time_bin_size=0.04).flux.min()

print(f"Odd dip: {b_odd:.5f} Even dip: {b_even:.5f}")
print(f"Rozdiel: {abs(b_odd-b_even)*100:.3f}% - ak <0.05% = OK!")

plt.figure(figsize=(8,3))
plt.scatter(fold_odd.phase.value, fold_odd.flux.value, s=1, c='blue', alpha=0.2, label='odd')
plt.scatter(fold_even.phase.value, fold_even.flux.value, s=1, c='red', alpha=0.2, label='even')
fold.bin(time_bin_size=0.04).scatter(s=20, c='black')
plt.xlim(-0.3,0.3); plt.ylim(0.992,1.004)
plt.legend()
plt.title(f"Odd-Even test {p:.4f} d")
plt.show()

depth = 1 - fold.bin(time_bin_size=0.04).flux.min()
noise = np.std(lc2_flat_fixed.flux.value)
n_points = len(fold[(abs(fold.phase.value)<0.1)].flux)
snr = depth/noise * np.sqrt(n_points)
print(f"SNR = {snr:.1f} - ak >10 = dobry!")
print(f"Pocet tranzitov cca {len(lc2_flat_fixed.time.value)/p:.0f}x")
```
